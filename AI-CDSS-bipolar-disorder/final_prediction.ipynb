```python
import pandas as pd
from datetime import datetime

# Load the datasets
interviews = pd.read_csv('interviews.csv')
episode_data = pd.read_csv('episode_data.csv')

# Convert date columns to datetime for comparison
interviews['Date'] = pd.to_datetime(interviews['Date'])
episode_data['Start_Date'] = pd.to_datetime(episode_data['Start_Date'])
episode_data['End_Date'] = pd.to_datetime(episode_data['End_Date'])

# Define the checkEpisode function
def checkEpisode(date, patient_code):
    # Filter episodes for the given patient
    patient_episodes = episode_data[episode_data['Patient_Code'] == patient_code]
    
    # Check if the date falls within any episode period
    for _, episode in patient_episodes.iterrows():
        if episode['Start_Date'] <= date <= episode['End_Date']:
            return episode['Episode_Type']
    return None  # Return None if no episode is found

# Create a copy of the interviews DataFrame
interviews_episodes = interviews.copy()

# Add the 'episode' column using correct column names
for index, row in interviews_episodes.iterrows():
    interviews_episodes.at[index, 'episode'] = checkEpisode(row['Date'], row['Patient_Code'])

# Save the updated DataFrame to a new CSV (optional)
interviews_episodes.to_csv('interviews_episodes.csv', index=False)

# Display the first few rows to verify
print(interviews_episodes.head())
```

```python
def check_state(classifier, data):
    prediction = classifier.predict(data)[0]
    if prediction == 'D':
        diagnosis = "The patient could be tending towards a Depression episode"
    elif prediction == 'M':
        diagnosis = "The patient could be tending towards a Mania episode"
    else:
        diagnosis = "The patient is in a Euthymic state"
    return diagnosis
```
